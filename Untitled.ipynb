```python
import os
import pandas as pd
from sqlalchemy import create_engine

# 1. Connection (note: postgresql+psycopg2)
engine = create_engine("postgresql+psycopg2://postgres:12345@localhost:5432/olist")
out_dir = os.path.expanduser("~/Documents/olist-project/dashboard/data")
os.makedirs(out_dir, exist_ok=True)

# 2. Customer segments
q = """
SELECT o.customer_unique_id, o.order_id, o.order_purchase_timestamp, r.revenue
FROM clean_orders o
JOIN order_revenue r ON r.order_id = o.order_id
"""
df = pd.read_sql(q, engine)
df["revenue"] = df["revenue"].astype(float)

snapshot = df["order_purchase_timestamp"].max() + pd.Timedelta(days=1)
rfm = df.groupby("customer_unique_id").agg(
    last_order=("order_purchase_timestamp", "max"),
    frequency=("order_id", "nunique"),
    monetary=("revenue", "sum"),
).reset_index()
rfm["recency"] = (snapshot - rfm["last_order"]).dt.days
rfm["R"] = pd.qcut(rfm["recency"], 5, labels=[5, 4, 3, 2, 1]).astype(int)
rfm["M"] = pd.qcut(rfm["monetary"].rank(method="first"), 5, labels=[1, 2, 3, 4, 5]).astype(int)

rfm["segment"] = "Lapsed one-timers"
rfm.loc[rfm["R"] >= 3, "segment"] = "Recent one-timers"
rfm.loc[(rfm["M"] >= 4) & (rfm["R"] >= 4), "segment"] = "New big spenders"
rfm.loc[rfm["frequency"] >= 2, "segment"] = "Repeat customers"
print(rfm["segment"].value_counts(normalize=True).round(3))

# 3. One order-level table for the Excel dashboard
q2 = """
SELECT o.order_id, o.customer_unique_id,
       DATE_TRUNC('month', o.order_purchase_timestamp)::date AS month,
       o.customer_state,
       o.is_late::int AS is_late,
       CASE WHEN NOT o.is_late THEN '1. On time'
            WHEN o.delay_days <= 5 THEN '2. Late 1-5 days'
            WHEN o.delay_days <= 10 THEN '3. Late 6-10 days'
            ELSE '4. Late 10+ days' END AS delay_group,
       v.review_score,
       r.revenue
FROM clean_orders o
JOIN order_revenue r ON r.order_id = o.order_id
LEFT JOIN clean_reviews v ON v.order_id = o.order_id
WHERE o.order_purchase_timestamp >= '2017-01-01'
  AND o.order_purchase_timestamp <  '2018-09-01'
"""
fact = pd.read_sql(q2, engine)
fact["revenue"] = fact["revenue"].astype(float)
fact = fact.merge(rfm[["customer_unique_id", "segment", "frequency"]],
                  on="customer_unique_id", how="left")
fact["customer_type"] = fact["frequency"].apply(lambda f: "Repeat" if f >= 2 else "One-time")
fact = fact.drop(columns=["frequency", "customer_unique_id"])

fact.to_csv(f"{out_dir}/orders_fact.csv", index=False)
print(fact.shape)
```

```python
import os
import pandas as pd
os.makedirs("dashboard_data", exist_ok=True)

queries = {
"monthly_revenue": """
    SELECT DATE_TRUNC('month', o.order_purchase_timestamp)::date AS month,
           COUNT(*) AS orders,
           ROUND(SUM(r.revenue)::numeric, 0) AS revenue
    FROM clean_orders o
    JOIN order_revenue r ON r.order_id = o.order_id
    WHERE o.order_purchase_timestamp >= '2017-01-01'
      AND o.order_purchase_timestamp <  '2018-09-01'
    GROUP BY 1 ORDER BY 1""",

"late_by_state": """
    SELECT customer_state, COUNT(*) AS orders,
           ROUND(100.0 * AVG(is_late::int), 1) AS late_pct
    FROM clean_orders
    GROUP BY customer_state
    HAVING COUNT(*) >= 200
    ORDER BY late_pct DESC""",

"review_by_delay": """
    SELECT CASE
             WHEN NOT o.is_late THEN '1. On time'
             WHEN o.delay_days <= 5 THEN '2. Late 1-5 days'
             WHEN o.delay_days <= 10 THEN '3. Late 6-10 days'
             ELSE '4. Late 10+ days' END AS delay_group,
           COUNT(*) AS orders,
           ROUND(AVG(v.review_score), 2) AS avg_review
    FROM clean_orders o
    JOIN clean_reviews v ON v.order_id = o.order_id
    GROUP BY 1 ORDER BY 1""",

"categories": """
    SELECT COALESCE(t.product_category_name_english, p.product_category_name, 'unknown') AS category,
           COUNT(DISTINCT oi.order_id) AS orders,
           ROUND(SUM(oi.price)::numeric, 0) AS revenue,
           ROUND(AVG(v.review_score), 2) AS avg_review,
           ROUND(100.0 * AVG(o.is_late::int), 1) AS late_pct
    FROM order_items oi
    JOIN clean_orders o ON o.order_id = oi.order_id
    JOIN clean_reviews v ON v.order_id = oi.order_id
    JOIN products p ON p.product_id = oi.product_id
    LEFT JOIN product_category_name_translation t
           ON t.product_category_name = p.product_category_name
    GROUP BY 1
    HAVING COUNT(DISTINCT oi.order_id) >= 300
    ORDER BY revenue DESC""",
}

for name, sql in queries.items():
    out = pd.read_sql(sql, engine)
    out.to_csv(f"dashboard_data/{name}.csv", index=False)
    print(name, out.shape)
```
